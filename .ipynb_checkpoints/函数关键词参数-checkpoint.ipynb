```python
def show_info(**pairs):
    print(pairs)        # 字典
    for key, value in pairs.items():
        print(f"{key} = {value}")
```

```python
show_info(stock = "aapl", price = 1000, num = 20)
```

```python
# 涉及几个参数是随意的
# 字典名叫pairs
# 你输入的都是字典里的键值对
# 函数的参数，可以是数字，也可以是对象
# 输入“对象”的：类型（数字、文本）/顺序/数量，各自有什么要求？
# print函数，无论打印出来是什么结果，本质都是单元元素的拼接，数字元素和文本元素的组合需要用逗号分隔开。
# print文本时需要“”引住，打印后不显示“”
# f- string formatted string literal，当编辑字符串中，需要添加变量时可以使用的方式
```

```python
def func(a,b,c,*num, **pair):
    print("a=",a,"b=",b,"c=",c)  
    print(f"a={a} b={b} c={c}")
    print(num)
    print(pair)
```

```python
func(1,2,3,4,4,5,color="blue")
```

```python
#设置了一个参数，但是使用时没有填入该参数值
#返回时会返回空值
#设置参数时顺序是必须时先普通参数*后默认参数**，填写值时填写顺序也不能颠倒
#它们让函数更灵活、能处理不确定数量和形式的参数。

def test(*price, **stock):
    print(price)
    print(stock)
```

```python
test(company = "aapl")
```
