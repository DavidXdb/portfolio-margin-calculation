```python
import csv

# --- 第 8 步修改开始 ---
# 新建一个空列表，用来存放从 CSV 读出的持仓。
positions = []
```

```python
with open("examples/portfolio_sample.csv", encoding="utf-8") as file:
    # csv.DictReader 把第一行当作列名，
    # 之后每一行变成一个字典：{列名: 该行的值}
    reader = csv.DictReader(file)
    
    for row in reader:
        # CSV 读出的值都是字符串（文本），不能直接计算，
        # 这里转换成正确的类型：
        #   price 是小数，用 float() 转换；
        #   quantity 和 multiplier 是整数，用 int() 转换。
        row["price"] = float(row["price"])
        row["quantity"] = int(row["quantity"])
        row["multiplier"] = int(row["multiplier"])

        # append() 把整理好的这条持仓加进列表的末尾
        positions.append(row)
        print(row)

print()
print(positions)
```

```python
# 返回结果分析：每个row都是一个字典。row["price"] 是每个row字典中price这个key对应的值；
# 返回结果分析：positions就是row（字典）组成的list
# 注意：前面的 Cell 里 with open(...) as file: 这个 with 块已经执行结束了，文件被自动关闭。
# 如果新开的 Cell 里却还在用 reader 去迭代，此时文件已关闭 → 报错。
# with 语句的作用就是：离开这个块时自动关闭文件。跨 Cell 继续用 reader 是无效的。
```

```python
securities = {}
with open("data/securities_info.csv", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        # 数值字段在加载时就转换好，之后所有使用方拿到的都是数字，
        # 避免每次使用时再转换。
        row["market_cap"] = float(row["market_cap"])
        row["adtv"] = float(row["adtv"])
        securities[row["ticker"]] = row

print(securities)
```

```python
# 上述代码理解：
# securities是一个字典，字典的key是公司ticker，key对应的值也是字典（证券的basic mrkt info）
# 字典securities的引用方式：
# 输入：securities["AAPL"]
# 返回：{'ticker': 'AAPL', 'name': 'Apple Inc.', 'country': 'USA', 'market_cap': 3000000000000.0, 'adtv': 50000000.0, 'gics_sector': '信息技术'}
```

```python
for position in positions:
    # 先判断该 ticker 是否存在于证券信息中。
    # "in" 用于检查字典里有没有某个键。
    # ES 是指数期货，没有单独的证券信息行，需要特殊处理。
    if position["ticker"] in securities:
        ref = securities[position["ticker"]]

        # 把证券信息字段逐个写入持仓字典。
        # market_cap 和 adtv 在加载证券信息时已转换为数字，这里直接复制。
        position["country"] = ref["country"]
        position["market_cap"] = ref["market_cap"]
        position["adtv"] = ref["adtv"]
        position["gics_sector"] = ref["gics_sector"]
        # Days of Holdings（持有天数）= 净股数的绝对值 ÷ 日均成交量。
        # 例如：持有 100 股、ADTV 为每天 5000 万股，
        # 全部卖出需要 100 ÷ 50000000 = 0.000002 天。
        
        days_of_holdings = abs(position["quantity"]) / position["adtv"]
        position["days_of_holdings"] = days_of_holdings

        # 判定 Issuer Grade（发行人等级）。
        # 数字中的下划线只是分隔符：10_000_000_000 就是 100 亿。
        market_cap = position["market_cap"]
        if market_cap >= 10_000_000_000 and days_of_holdings < 0.5:
            issuer_grade = "High Grade"
        elif market_cap >= 3_000_000_000 and days_of_holdings < 1:
            issuer_grade = "Medium Grade"
        else:
            issuer_grade = "Low Grade"
        position["issuer_grade"] = issuer_grade

        # 判定是否属于 Large Cap, Liquid Security：
        # 市值大于 100 亿，且持有天数小于 1 天。
        if market_cap > 10_000_000_000 and days_of_holdings < 1:
            large_cap_liquid = True
        else:
            large_cap_liquid = False
            
        position["large_cap_liquid"] = large_cap_liquid
        
```

```python
# 新增了position下面 ["issuer_grade"]，["days_of_holdings"]，["large_cap_liquid"] 这几个key
```

```python
# 上述代码理解：
# positions 是list；position 是字典
# position["ticker"] 读的是每个position 中 ticker这个key查询出来的值
# securities 也是一个用每个公司的ticker当key来查询的字典，所以可以写 if position["ticker"] in securities
# 本质等于，if 'AAPL' in securities
# ref 是什么？ref是字典。
# ref的一个示例：{'ticker': 'AAPL', 'name': 'Apple Inc.', 'country': 'USA', 'market_cap': 3000000000000.0, 'adtv': 50000000.0, 'gics_sector': '信息技术'}
```

```python
# 本质上：通过ref（字典）这个中介，把securities（由字典组成的字典）中的信息读出来，写入positions（列表）中的每一个position（字典）
```

```python
constituents = []
with open("data/index_constituents.csv", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        row["weight"] = float(row["weight"])
        constituents.append(row)
```

```python
constituent_positions = []
for position in positions:
    if position["ticker"] == "ES":
        # 期货的名义价值 = 价格 × 数量 × 乘数
        index_value = position["price"] * position["quantity"] * position["multiplier"]

        for constituent in constituents:
            # 该成份的市值份额 = 名义价值 × 权重
            value = index_value * constituent["weight"]

            constituent_positions.append({
                "ticker": constituent["constituent_ticker"],
                "value": value,
                "direction": position["direction"],
            })
```

```python
# constituents 也是由字典组成的列表list
# 新增的 constituent_positions 也是由字典组成的列表list，其中每个字典有三个key"ticker"，"value"，"direction"
# 每一个constituent都有对应的值：value = index_value * constituent["weight"]
# value被添加到 constituent_positions 中
```

```python
long_total = 0
short_total = 0

for position in positions:

    # 市值 = 当日价格 × 持仓数量 × 标的乘数
    market_value = position["price"] * position["quantity"] * position["multiplier"]

    # 根据头寸方向确定符号：多头 +1，空头 -1
    if position["direction"] == "多头":
        sign = 1
    elif position["direction"] == "空头":
        sign = -1
    else:
        # 方向既不是"多头"也不是"空头"，说明数据有问题。
        # raise 表示主动报错并停止程序，避免用错误数据继续计算。
        raise ValueError("未知的头寸方向：" + position["direction"])

    # 有向市值 = 市值 × 符号
    signed_market_value = market_value * sign
    
    position["market_value"] = market_value
    position["signed_market_value"] = signed_market_value
    
    if position["direction"] == "多头":
        long_total = long_total + signed_market_value
    else:
        short_total = short_total + signed_market_value
    
```

```python
issuer_exposure = {}

# 先计算直接持仓中的敞口（ES 除外，它已由成份持仓代表）。
for position in positions:
    if position["ticker"] == "ES":
        # continue：跳过本次循环的剩余部分，直接进入下一轮
        continue
        
    ticker = position["ticker"]
    if ticker not in issuer_exposure:
        issuer_exposure[ticker] = 0
        
    issuer_exposure[ticker] = issuer_exposure[ticker] + position["signed_market_value"]

print(issuer_exposure)
```

```python
# 直接赋值写法：issuer_exposure[ticker] =  position["signed_market_value"]
# 直接赋值写法的问题，如果持仓中每个 ticker 只出现一次时，该写法才是对的
# 对于每个position，先检查ticker是否为ES，否则跳过
# 读取每个position的ticker，写一个新变量 ticker = position["ticker"]
# 对于每个position的ticker，如果字典issuer_exposure 的key中没有该ticker，则写入，新建值为零
# 同样还是对于该position的ticker，issuer_exposure[ticker] 等于之前的已有值加上该position的signed_value
```

```python
# 再加成份持仓（方向与 ES 持仓一致）。
for constituent_position in constituent_positions:
    ticker = constituent_position["ticker"]
    
    if constituent_position["direction"] == "多头":
        signed_value = constituent_position["value"]
    else:
        signed_value = -constituent_position["value"]

    if ticker not in issuer_exposure:
        issuer_exposure[ticker] = 0
    issuer_exposure[ticker] = issuer_exposure[ticker] + signed_value

print(issuer_exposure)
```

```python
largest_short_stress = 0

for ticker, net in issuer_exposure.items():
    if net < 0:
        # 查该发行人的市值（证券信息在加载时已转为数字）
        market_cap = securities[ticker]["market_cap"]

        # 按市值档位确定压力百分比 Y
        if market_cap < 100_000_000:          # 小于 1 亿
            stress = 1000
        elif market_cap < 250_000_000:        # 1 亿（含）到 2.5 亿
            stress = 750
        elif market_cap < 5_000_000_000:      # 2.5 亿（含）到 50 亿
            stress = 500
        elif market_cap < 10_000_000_000:     # 50 亿（含）到 100 亿
            stress = 300
        elif market_cap < 15_000_000_000:     # 100 亿（含）到 150 亿
            stress = 200
        elif market_cap < 20_000_000_000:     # 150 亿（含）到 200 亿
            stress = 150
        else:                                 # 200 亿及以上
            stress = 100

        # 压力损失 = |净空头市值| × Y / 100
        loss = abs(net) * stress / 100

        print(ticker, "净空头市值 =", net, "压力 =", stress, "% 损失 =", loss)

        # 与目前的最大值比较，更大就替换
        if loss > largest_short_stress:
            largest_short_stress = loss

print("最大 Short Issuer Stress 损失 =", largest_short_stress)
print()
```

```python
# Top 5 要求：
# 对每个发行人分别做 +50% 与 -50% 的独立压力测试。
# 线性头寸下，两种情景中的最大损失都是 |净敞口| × 50%：
#   多头在下跌 50% 时损失，空头在上涨 50% 时损失，金额相同。
# 取损失最大的 5 个发行人，求和。
stress_losses = []
for ticker, net in issuer_exposure.items():
    loss = abs(net) * 0.5
    # 元组 (损失, ticker)：损失放在前面，便于排序
    stress_losses.append((loss, ticker))

# sort(reverse=True)：从大到小排序。
# 元组排序先比第一个元素（损失），损失相同时再比 ticker。
stress_losses.sort(reverse=True)

# 切片 [:5] 取前 5 个；不足 5 个时取全部
top_5_requirement = 0
for loss, ticker in stress_losses[:5]:
    print(ticker, "压力损失 =", loss)
    top_5_requirement = top_5_requirement + loss

print("Top 5 要求 =", top_5_requirement)
print()
```

```python
# 循环结束后，打印组合层面的两个合计值。
# 注意：这里的语句不带缩进，属于循环外，只会执行一次。
print()
print("多头合计 =", long_total)
print("空头合计 =", short_total)

# 净敞口 = 多头合计 + 空头合计。空头合计本身是负数，两者相加就是"多空相抵"。
# 毛市值 = 多头合计的绝对值 + 空头合计的绝对值。毛市值不冲抵：多头 465000 加空头侧规模 10000，共 475000。
# abs() 是 Python 内置函数，用来取绝对值。

net_exposure = long_total + short_total
gross_market_value = abs(long_total) + abs(short_total)

print("净敞口 =", net_exposure)
print("毛市值 =", gross_market_value)
```

```python
# --- 第 17 步新增开始 ---
# 分层计收：
# 10 亿以内部分按 12.5%；
# 超出 10 亿的增量部分按 35%。
# 注意：35% 只针对超出部分，10 亿以内仍按 12.5%。

if gross_market_value <= 1_000_000_000:
    section_a_part_a = gross_market_value * 12.5 / 100
else:
    section_a_part_a = 1_000_000_000 * 12.5 / 100
    section_a_part_a = section_a_part_a + (gross_market_value - 1_000_000_000) * 35 / 100
print("Section A a) 分层要求 =", section_a_part_a)

# b) Long/Short 比率
# 比率 = 多头毛市值 ÷ 空头毛市值。
# 空头合计存的是带符号的负数，这里用 abs() 转成正数再相除。
# 比率 > 1.05 或 < 0.95 时，加收 20% × |净敞口|

long_short_ratio = long_total / abs(short_total)
print("Long/Short 比率 =", long_short_ratio)

if long_short_ratio > 1.05 or long_short_ratio < 0.95:
    section_a_part_b = abs(net_exposure) * 20 / 100
else:
    section_a_part_b = 0
print("Section A b) 比率要求 =", section_a_part_b)

section_a = section_a_part_a + section_a_part_b

print("Section A 初始要求 =", section_a)
print()
```

```python
# 新建的字典是跳出持仓遍历外的，持仓中有指标A不唯一，我们需要针对A进行统计（例如求和），
# 那就新建字典，针对每个持仓都问一遍他们的指标A在不在字典里，如果不在就新增。新增之后添加值
# 这不就是groupby吗?
```

```python
# --- 第 19 步新增开始（一）单一国家条件 ---
# 读取国家档位文件：{country: tier}
country_tiers = {}
with open("data/country_tiers.csv", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        country_tiers[row["country"]] = row["tier"]

# 读数取数操作。
# row是字典。key有两个，都是抽象的类别：country 以及 tier
# country_tiers 是字典，key是具体的国家，如：US，UK，国家有多少个，key就有多少个

# 按国家汇总净敞口：{country: 该国发行人净有向市值之和}
country_nmv = {}
for ticker, net in issuer_exposure.items():
    country = securities[ticker]["country"]
    if country not in country_nmv:
        country_nmv[country] = 0
    country_nmv[country] = country_nmv[country] + net

# 字典中字典的取值方式：securities[ticker]["country"]
# 每个ticker都要进行下列操作：
# 如果判断：某个ticker读出的国家不在字典country_nmv的key中，则存入并写入初始值country_nmv[country] = 0；存过就不用新增
# if分类判断之后，把当前ticker对应的net值加进来

# 单一国家条件：|该国净敞口| 超过"阈值 × 毛市值"的增量部分计收
single_country_requirement = 0
for country, nmv in country_nmv.items():
    tier = country_tiers[country]
    if tier == "Tier 1":
        threshold_rate = 10 / 100
        charge_rate = 10 / 100
    elif tier == "Tier 2":
        threshold_rate = 5 / 100
        charge_rate = 10 / 100
    elif tier == "Tier 3":
        threshold_rate = 5 / 100
        charge_rate = 15 / 100
    else:
        threshold_rate = 2.5 / 100
        charge_rate = 15 / 100

    threshold_amount = threshold_rate * gross_market_value
    excess = abs(nmv) - threshold_amount
    if excess > 0:
        requirement = excess * charge_rate
    else:
        requirement = 0
    print(country, tier, "净敞口 =", nmv, "阈值金额 =", threshold_amount, "附加要求 =", requirement)
    single_country_requirement = single_country_requirement + requirement
    
# 并入 Section A（国家附加要求属于 Section A，参与取大比较）
section_a = section_a + single_country_requirement
print("单一国家附加要求合计 =", single_country_requirement)
print("Section A 累计 =", section_a)
print()

# 发行人层面毛市值归集：直接持仓与成份持仓按 |市值| 相加
issuer_gmv = {}
for position in positions:
    if position["ticker"] == "ES":
        continue
    ticker = position["ticker"]
    if ticker not in issuer_gmv:
        issuer_gmv[ticker] = 0
    issuer_gmv[ticker] = issuer_gmv[ticker] + abs(position["signed_market_value"])

for constituent_position in constituent_positions:
    ticker = constituent_position["ticker"]
    if ticker not in issuer_gmv:
        issuer_gmv[ticker] = 0
    issuer_gmv[ticker] = issuer_gmv[ticker] + constituent_position["value"]

# 验证：归集合计应等于组合毛市值
gmv_total = 0
for ticker, gmv in issuer_gmv.items():
    gmv_total = gmv_total + gmv
print("发行人毛市值归集合计 =", gmv_total)

# 有直接持仓的发行人，直接沿用第 12 步算好的 LCL 标志
direct_lcl = {}
for position in positions:
    if position["ticker"] == "ES":
        continue
    direct_lcl[position["ticker"]] = position["large_cap_liquid"]

# 国家汇总条件：Tier 2 一律 2.5%；Tier 3/4 区分 LCL 与非 LCL；Tier 1 无
aggregate_country_requirement = 0
for ticker, gmv in issuer_gmv.items():
    tier = country_tiers[securities[ticker]["country"]]
    if tier == "Tier 1":
        continue

    if ticker in direct_lcl:
        lcl = direct_lcl[ticker]
    else:
        # 纯成份发行人缺少股票价格，持有天数近似为 0，仅按市值判定
        lcl = securities[ticker]["market_cap"] > 10_000_000_000

    if tier == "Tier 2":
        rate = 2.5 / 100
    elif tier == "Tier 3" and lcl:
        rate = 10 / 100
    elif tier == "Tier 3":
        rate = 15 / 100
    elif tier == "Tier 4" and lcl:
        rate = 15 / 100
    else:
        rate = 20 / 100

    requirement = gmv * rate
    print(ticker, tier, "毛市值 =", gmv, "要求 =", requirement)
    aggregate_country_requirement = aggregate_country_requirement + requirement

# 并入 Section A
section_a = section_a + aggregate_country_requirement
print("国家汇总附加要求合计 =", aggregate_country_requirement)
print("Section A 累计 =", section_a)
print()
```

```python
# --- 第 20 步新增开始 ---
# 流动性附加要求：按持有天数分档，对"超过档位下限天数"的部分计收。
# 分档：>0.2 至 ≤1 天 15%；>1 至 ≤2 天 35%；>2 至 ≤3 天 50%。
# 各档独立判断（用 if 而不是 elif）：持有 2.5 天时三档同时计收。
# 成份持仓缺少价格与股数数据，本步暂不处理（样本持有天数远小于 0.2）。
liquidity_requirement = 0
for position in positions:
    if position["ticker"] == "ES":
        continue

    days_of_holdings = position["days_of_holdings"]
    market_value = position["market_value"]

    charge = 0
    if days_of_holdings > 0.2:
        # 超过 0.2 天的部分（以 1 天为上限）
        excess_days = min(days_of_holdings, 1) - 0.2
        charged_value = market_value * excess_days / days_of_holdings
        charge = charge + charged_value * 15 / 100

    if days_of_holdings > 1:
        excess_days = min(days_of_holdings, 2) - 1
        charged_value = market_value * excess_days / days_of_holdings
        charge = charge + charged_value * 35 / 100

    if days_of_holdings > 2:
        excess_days = min(days_of_holdings, 3) - 2
        charged_value = market_value * excess_days / days_of_holdings
        charge = charge + charged_value * 50 / 100

    print(position["ticker"], "DoH =", days_of_holdings, "流动性要求 =", charge)
    liquidity_requirement = liquidity_requirement + charge

# 并入 Section A
section_a = section_a + liquidity_requirement
print("流动性附加要求合计 =", liquidity_requirement)
print("Section A 累计 =", section_a)
print()
```

```python
# --- 第 21 步新增开始 ---
# 行业附加要求：单一 GICS 行业的 |净敞口| 超过毛市值的 15% 时，
# 对增量部分计收 15%。
sector_nmv = {}
for ticker, net in issuer_exposure.items():
    sector = securities[ticker]["gics_sector"]
    if sector not in sector_nmv:
        sector_nmv[sector] = 0
    sector_nmv[sector] = sector_nmv[sector] + net

# 与单一国家条件不同：行业只有一档（15%），阈值金额只需算一次
sector_requirement = 0
threshold_amount = 15 / 100 * gross_market_value
for sector, nmv in sector_nmv.items():
    excess = abs(nmv) - threshold_amount
    if excess > 0:
        requirement = excess * 15 / 100
    else:
        requirement = 0
    print(sector, "净敞口 =", nmv, "阈值金额 =", threshold_amount, "附加要求 =", requirement)
    sector_requirement = sector_requirement + requirement

# 并入 Section A
section_a = section_a + sector_requirement
print("行业附加要求合计 =", sector_requirement)
print("Section A 累计 =", section_a)
print()
```

```python
# --- section A合计（第 18 步新增）---
# Base Margin Requirement：三个比较项取最大。
# max() 是内置函数，返回多个数值中最大的一个。

base_margin_requirement = max(section_a, largest_short_stress, top_5_requirement)

print("Base Margin Requirement =", base_margin_requirement)
print()
```

```python
# --- 第 23 步新增开始 ---
# 总保证金 = Base Margin + Portfolio Conditions
total_margin_requirement = base_margin_requirement + portfolio_conditions_requirement

# 汇总清单：把此前分散打印的结果集中呈现
print("===== 保证金汇总 =====")
print("Section A 初始要求 =", section_a_part_a + section_a_part_b)
print("  其中 a) 毛市值分层 =", section_a_part_a)
print("  其中 b) Long/Short 比率 =", section_a_part_b)
print("国家附加要求 =", single_country_requirement + aggregate_country_requirement)
print("  其中 单一国家条件 =", single_country_requirement)
print("  其中 国家汇总条件 =", aggregate_country_requirement)
print("流动性附加要求 =", liquidity_requirement)
print("行业附加要求 =", sector_requirement)
print("Section A 合计 =", section_a)
print("最大 Short Issuer Stress 损失 =", largest_short_stress)
print("Top 5 要求 =", top_5_requirement)
print("Base Margin Requirement =", base_margin_requirement)
print("Portfolio Conditions =", portfolio_conditions_requirement)
print("总保证金 =", total_margin_requirement)
print("===== 汇总结束 =====")
print()
```
