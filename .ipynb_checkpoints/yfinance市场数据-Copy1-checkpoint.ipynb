```python
import yfinance as yf
from curl_cffi import requests
import pandas as pd
from concurrent.futures import ThreadPoolExecutor, as_completed
import time
import os
import random
import math
```

```python
_YF_SESSION = requests.Session(impersonate="chrome")
yf.config.network.proxy = "http://127.0.0.1:7890"
```

```python
def bbg_to_key(ticker: str) -> str:
    """'688807 CH Equity' -> '688807'（仅取代码段并补零，用于查 group_map）"""
    return ticker.split()[0].strip().zfill(6)


def retry(func, *args, retries=3, backoff=1.0, **kwargs):
    for i in range(retries):
        try:
            return func(*args, **kwargs)
        except Exception:
            if i == retries - 1:
                raise
            time.sleep(backoff * (2 ** i) + random.uniform(0, 0.5))
```

```python
# ============ 本地映射：group 和 y_ticker ============

def load_group_map(path: str = "group_map.csv"):
    """
    读取 group_map.csv（列：ticker, group, y_ticker），
    返回 (group_map, yf_map)：{6位代码: group} 和 {6位代码: y_ticker}。
    """
    df = pd.read_csv(path, dtype={"ticker": str, "y_ticker": str})
    df["ticker"] = (
        df["ticker"].astype(str).str.strip()
        .str.replace(r"\.0$", "", regex=True)
        .str.zfill(6)
    )
    df = df.dropna(subset=["ticker", "y_ticker"])
    group_map = dict(zip(df["ticker"], df["group"]))
    yf_map = dict(zip(df["ticker"], df["y_ticker"]))
    print(f"已加载映射 {len(group_map)} 条")
    return group_map, yf_map
    
```

```python
# ============ 逐只获取其余指标 ============

def get_market_cap_one(yf_ticker: str):
    info = retry(lambda: yf.Ticker(yf_ticker, session=_YF_SESSION).info, retries=3)
    return info.get("marketCap")


def get_adtv_one(yf_ticker: str, days: int = 15):
    def _fetch():
        hist = yf.Ticker(yf_ticker, session=_YF_SESSION).history(period=f"{days * 2}d", auto_adjust=False)
        if hist.empty or "Volume" not in hist.columns:
            return None
        vol = hist["Volume"].dropna().tail(days)
        if vol.empty:
            return None
        return float(vol.mean())  # 单位：股；如需「手」则 /100
    return retry(_fetch, retries=3)


def fetch_one(ticker: str, adtv_days: int, group_map: dict, yf_map: dict):
    """抓单只股票；单只失败不抛出，返回带 error 的行"""
    key = bbg_to_key(ticker)
    yf_ticker = yf_map.get(key)
    if not yf_ticker:
        return {
            "ticker": ticker, "symbol": key,
            "error": "y_ticker not found in group_map",
        }

    row = {
        "ticker": ticker,
        "symbol": key,
        "y_ticker": yf_ticker,
        "sector": group_map.get(key),
    }

    try:
        row["market_cap"] = get_market_cap_one(yf_ticker)
    except Exception as e:
        row["market_cap"] = None
        row["error_market_cap"] = str(e)

    try:
        row[f"adtv_{adtv_days}d"] = get_adtv_one(yf_ticker, adtv_days)
    except Exception as e:
        row[f"adtv_{adtv_days}d"] = None
        row[f"error_adtv_{adtv_days}d"] = str(e)

    time.sleep(random.uniform(2, 4))
    return row
```

```python
# ============ 主流程：并发 + 断点续跑 ============

def build_dataset(
    bbg_tickers: list,
    adtv_days: int = 20,
    max_workers: int = 2,
    cache_path: str = "yf_test.csv",
    group_map_path: str = "group_map.csv",
):
    group_map, yf_map = load_group_map(group_map_path)

    done = {}
    if os.path.exists(cache_path):
        cached = pd.read_csv(cache_path)
        done = {row["ticker"]: row.to_dict() for _, row in cached.iterrows()}
        print(f"已加载缓存 {len(done)} 条")

    todo = [t for t in bbg_tickers if t not in done]
    print(f"待处理 {len(todo)} 个 ticker")
    if not todo:
        return pd.DataFrame(done.values())

    rows = []
    with ThreadPoolExecutor(max_workers=max_workers) as pool:
        futures = {pool.submit(fetch_one, t, adtv_days, group_map, yf_map): t for t in todo}
        for i, fut in enumerate(as_completed(futures), 1):
            t = futures[fut]
            try:
                rows.append(fut.result())
            except Exception as e:
                print(f"[失败] {t}: {e}")
                rows.append({"ticker": t})
            if i % 50 == 0:
                _flush(rows, cache_path)
                print(f"进度 {i}/{len(todo)}")

    _flush(rows, cache_path)
    all_rows = list(done.values()) + rows
    return pd.DataFrame(all_rows)


def _flush(rows, cache_path):
    if not rows:
        return
    new_df = pd.DataFrame(rows)
    if os.path.exists(cache_path):
        old_df = pd.read_csv(cache_path)
        new_df = pd.concat([old_df, new_df], ignore_index=True)
    new_df.drop_duplicates(subset="ticker", keep="last").to_csv(cache_path, index=False)
```

```python
if __name__ == "__main__":

    portfolio_df = pd.read_csv("/Users/bigstar/pb-margin-calculator/examples/portfolio_sample.csv", 
                               encoding="gbk",)
    bbg_tickers = portfolio_df["ticker"].dropna().drop_duplicates().tolist()
    
    batch_size = 100
    i = 16  
    # 每次手动改这个：0, 1, 2, ... 16
    
    batch = bbg_tickers[i * batch_size : (i + 1) * batch_size]
    print(f"第 {i+1} 组，共 {len(batch)} 个 ticker")
    
    df = build_dataset(
        batch,
        adtv_days=15,
        max_workers=2,
        cache_path="yf_test.csv",
        group_map_path="group_map.csv",
    )
    
    print(df.head())
    print(f"总行数: {len(df)}")
```
